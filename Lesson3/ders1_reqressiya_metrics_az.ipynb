Machine Learning (ML) üç əsas qrupa bölünür. Bu dərs **Supervised Learning** üzərindədir, amma digər ikisini də tanımaq lazımdır.

| Növ | Target (cavab) varmı? | Öyrənmə necədir? | Nümunə |

### 🎓 Supervised Learning (Nəzarətli öyrənmə)

Təsəvvür edin ki, uşağa şəkillər göstərib deyirsiniz: *"bu pişikdir, bu itdir"*. Uşaq nümunələrə baxıb öyrənir.

**Supervised Learning də belədir:** kompüterə həm **giriş verilənlərini (Features)**, həm də **düzgün cavabları (Target)** veririk.

| Features (X) | Target (y) |
|---|---|
| Mənzilin sahəsi | Mənzilin qiyməti |
| Şəkil | "pişik" / "it" |
| E-poçt mətni | "spam" / "not spam" |

Model nümunələrə baxır, qanunauyğunluğu tapır və **yeni** suallara cavab verə bilir.

İki əsas tapşırıq:
- **Classification** — cavab kateqoriyadır
- **Regression** — cavab ədəddir ← *bu dərsin mövzusu*

### 🔍 Unsupervised Learning (Nəzarətsiz öyrənmə)

Burada **Target yoxdur** — yalnız Features veririk. Model özü verilənlərdə **qrup** və ya **struktur** axtarır.

Analogiya: uşağa şəkillər verirsiniz, amma ad demirsiniz. Uşaq özü oxşar şəkilləri bir yerə yığır: *"bunlar bir-birinə oxşayır, onlar ayrıdır"*.

| Nümunə | Nə edir? |
|---|---|
| 🛒 Market müştəriləri | Alış-verişə görə qruplara ayırır (seqmentasiya) |
| 📰 Xəbər mətnləri | Oxşar mövzuları bir yerə yığır (klasterləşdirmə) |
| 📉 Anomaliya aşkarlama | "Normal"dan kənar nöqtələri tapır |

Ən məşhur alqoritmlər: **K-Means**, **Hierarchical Clustering**, **PCA** (ölçü azaltma).

> 💡 Bu dərsdə Unsupervised yoxdur — gələcək dərslərdə öyrənəcəyik.

### 🎮 Reinforcement Learning (Möhkəmləndirmə ilə öyrənmə)

Agent **ətraf mühitdə** hərəkət edir, hər addımdan sonra **mükafat (reward)** və ya **cəza** alır. Məqsəd: uzun müddətdə mükafatı maksimallaşdırmaq.

Real həyatda:
- ♟️ AlphaGo / şahmat AI
- 🤖 Robotların yeriməyi öyrənməsi

> 💡 Supervised-dən fərq: hazır "düzgün cavab" yoxdur — agent özü sınaqla öyrənir.

## 3️⃣ Regression anlayışı (Concept of Regression)

**Regression = verilənlərə ən yaxşı uyğun gələn xətti (qanunauyğunluğu) tapmaq.**

Məsələ: *Mənzilin sahəsi artdıqca qiyməti necə dəyişir?*

Əvvəlcə kitabxanaları yükləyək.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

plt.rcParams["figure.figsize"] = (7, 4.5)
plt.rcParams["font.size"] = 12
````

### Dataset
Kiçik, **illüstrativ** (uydurma) cədvəl: 10 mənzilin sahəsi və qiyməti.

````python
df = pd.DataFrame({
    "sahe":   [35, 45, 50, 60, 65, 75, 85, 95, 110, 130],      # kv.m
    "qiymet": [75, 95, 100, 130, 125, 150, 175, 185, 215, 255]  # min AZN
})
df
````

Rəqəmlərə baxmaq çətindir. **Qrafikə** (scatter plot) baxaq 👇

````python
plt.scatter(df["sahe"], df["qiymet"], s=70)
plt.xlabel("Sahə (kv.m)")
plt.ylabel("Qiymət (min AZN)")
plt.title("Mənzillərin sahəsi və qiyməti")
plt.grid(alpha=0.3)
plt.show()
````

### 👀 Qrafikdən nə görürük?

Nöqtələr təxminən **düz xətt boyunca** düzülüb: sahə artdıqca qiymət də artır.

**Regression-un məqsədi:** bu nöqtələrin arasından keçən ən yaxşı xətti tapmaq. Sonra həmin xətt vasitəsilə **yeni mənzilin qiymətini** təxmin edə bilərik (prediction).

Ancaq "ən yaxşı xətt" nə deməkdir? Gəlin özümüz cəhd edək.

## 4️⃣ Bir dəyişənli Regression (Simple Linear Regression)

### Xəttin "resepti": iki ədəd
Hər düz xətti **iki ədəd** təyin edir. Taksi nümunəsinə baxaq:

> **Taksi haqqı = başlanğıc haqq + hər km üçün qiymət × məsafə**

- Başlanğıc haqq: **1.5 AZN** (taksiyə oturan kimi yazılır)
- Hər km: **0.6 AZN**

````python
def taksi_qiymeti(km):
    baslangic = 1.5      # b (bias): başlanğıc haqq
    her_km = 0.6         # w (weight): hər km-də qiymət nə qədər artır
    return baslangic + her_km * km

for km in [0, 5, 10, 20]:
    print(f"{km:>2} km -> {taksi_qiymeti(km):.1f} AZN")
````

### 🔑 Regression-un iki sehrli ədədi

Mənzil üçün də eyni məntiq:

> **qiymət = w × sahə + b**

| Ədəd | Adı | Mənası |
|---|---|---|
| **w** | Weight (çəki) | Sahə 1 kv.m artanda qiymət nə qədər artır? |
| **b** | Bias (sürüşmə) | Sahə 0 olanda başlanğıc qiymət |

**Regression-da "öyrənmək" (training) = ən yaxşı w və b ədədlərini tapmaq.** Hamısı budur!

### 🎯 Əl ilə cəhd edək

Üç fərqli "təxmin" xətti çəkək. Hansı daha yaxşı görünür?

````python
x_xett = [30, 135]    # xətti çəkmək üçün iki nöqtə kifayətdir

plt.scatter(df["sahe"], df["qiymet"], s=70, color="black", label="Real mənzillər")

# (w, b, rəng, ad)
texminler = [
    (1.0, 50, "red",    "Xətt A: w=1, b=50"), # formula y = wx + b
    (3.0, 0,  "orange", "Xətt B: w=3, b=0"),
    (2.0, 5,  "green",  "Xətt C: w=2, b=5"),
]
for w, b, reng, ad in texminler:
    plt.plot(x_xett, [w * x + b for x in x_xett], color=reng, label=ad)

plt.xlabel("Sahə (kv.m)")
plt.ylabel("Qiymət (min AZN)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

**Sual:** Sizcə hansı xətt ən yaxşıdır? Niyə?

Çox güman ki, **C** deyəcəksiniz, çünki nöqtələrə ən yaxın keçir.

### 📏 Error = xəttin nöqtədən nə qədər uzaq olması

Hər mənzil üçün **real qiymət** ilə **xəttin dediyi qiymət (prediction)** arasındakı fərqə **error** deyirik.

Xətt C üçün error-ları qrafikdə göstərək (qırmızı xətlər):

````python
w, b = 2.0, 5
texmin = w * df["sahe"] + b          # xəttin dediyi qiymətlər (prediction)

plt.scatter(df["sahe"], df["qiymet"], s=70, color="black", label="Real qiymət")
plt.plot(x_xett, [w * x + b for x in x_xett], color="green", label="Xətt C")

# Hər nöqtədən xəttə qırmızı xətt çəkirik
for x, real, tex in zip(df["sahe"], df["qiymet"], texmin):
    plt.plot([x, x], [real, tex], color="red", linewidth=2)

plt.xlabel("Sahə (kv.m)")
plt.ylabel("Qiymət (min AZN)")
plt.title("Qırmızı xətlər = error")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

### 🧮 Üç xətti rəqəmlə müqayisə edək

Hər xətt üçün **orta error** hesablayırıq (işarəsiz). Nə qədər kiçikdirsə, xətt bir o qədər yaxşıdır.

````python
for w, b, reng, ad in texminler:
    texmin = w * df["sahe"] + b
    error = abs(df["qiymet"] - texmin)     # hər mənzil üçün error
    print(f"{ad:20s} -> orta error = {error.mean():.1f} min AZN")
````

✅ **Xətt C-nin error-u ən kiçikdir.** Deməli, "ən yaxşı xətt" = **ən kiçik error-u olan xətt**.

Amma w və b üçün milyonlarla ehtimalı əl ilə yoxlamaq mümkün deyil. **Kompüter bunu bizim üçün avtomatik edir.** 👇

### 🤖 Kompüter ən yaxşı xətti özü tapır

`scikit-learn` kitabxanasında `LinearRegression` modeli var. Üç addım:

1. Modeli yaradırıq
2. `fit` ilə **öyrədirik** (ən yaxşı w və b-ni tapır)
3. `predict` ilə **prediction** veririk

````python
X = df[["sahe"]]        # Features (X)
y = df["qiymet"]        # Target (y)

model = LinearRegression()      # 1. model yaradırıq
model.fit(X, y)                 # 2. öyrədirik

w = model.coef_[0]
b = model.intercept_
print(f"Model tapdı: w = {w:.2f},  b = {b:.2f}")
print(f"Yəni: qiymət = {w:.2f} × sahə + {b:.2f}")
````

**Mənası:** hər əlavə 1 kv.m sahə qiyməti təxminən **w** min AZN artırır. Bizim əl ilə seçdiyimiz Xətt C-yə (w=2, b=5) çox yaxındır!

````python
plt.scatter(df["sahe"], df["qiymet"], s=70, color="black", label="Real mənzillər")
plt.plot(x_xett, [w * x + b for x in x_xett], color="blue", linewidth=3, label="Modelin tapdığı xətt")
plt.xlabel("Sahə (kv.m)")
plt.ylabel("Qiymət (min AZN)")
plt.title("Öyrədilmiş Regression xətti")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

### 🔮 Prediction: yeni mənzilin qiyməti

Modelin heç vaxt görmədiyi **100 kv.m**-lik mənzil üçün qiymət təxmin edək.

````python
yeni = pd.DataFrame({"sahe": [100]})
proqnoz = model.predict(yeni)
print(f"100 kv.m mənzil üçün prediction: {proqnoz[0]:.0f} min AZN")
````

````python
# Bir neçə mənzil birdən
yeniler = pd.DataFrame({"sahe": [40, 70, 100, 120]})
yeniler["prediction"] = model.predict(yeniler).round(0)
yeniler
````

### ⚠️ Diqqət: model yalnız gördüyü verilənlər qədər ağıllıdır

Bizim verilənlərdə mənzillər 35-130 kv.m arasındadır. Bəs 500 kv.m üçün?

````python
print("500 kv.m ->", round(model.predict(pd.DataFrame({"sahe": [500]}))[0]), "min AZN")
````

Model bir rəqəm verir, amma **ona etibar etmək olmaz**: 500 kv.m-lik mənzilləri heç görməyib.

> 💡 **Qayda:** model yalnız öyrəndiyi diapazonda etibarlıdır.

## 5️⃣ Üç növ Regression modeli

Regression-un bir neçə növü var. Bu dərsdə **ən çox istifadə olunan üçünü** tanıyırıq:

| № | Model | Neçə feature? | Forma | Formula |
|---|---|---|---|---|
| 1 | **Simple Linear Regression** | 1 | Düz xətt | $y = w x + b$ |
| 2 | **Multiple Linear Regression** | 2 və ya daha çox | Düz "hipermüstəvi" | $y = w_1 x_1 + w_2 x_2 + \ldots + b$ |
| 3 | **Polynomial Regression** | 1 və ya daha çox | Əyri (parabola və s.) | $y = w_1 x + w_2 x^2 + \ldots + b$ |

> 💡 Hamısı hələ də **"linear"** adlanır — çünki model **w və b-yə görə** xəttidir. Əyri görünən Polynomial-da belə, kompüter yenə w-ləri düz xətt kimi öyrənir.

### 1️⃣ Simple Linear Regression (Bir dəyişənli)

Bunu artıq öyrəndik: **bir** feature → **bir** target.

> **qiymət = w × sahə + b**

- ✅ Sadədir, oxumaq asandır
- ✅ Feature ilə target arasında **düz xətt** əlaqə varsa yaxşı işləyir
- ❌ Real həyatda çox vaxt bir feature kifayət etmir (məsələn, mənzil qiymətinə yalnız sahə deyil, mərtəbə, rayon da təsir edir)

### 2️⃣ Multiple Linear Regression (Çox dəyişənli)

İndi **bir neçə** feature eyni anda istifadə olunur:

> **qiymət = w₁ × sahə + w₂ × otaq + w₃ × mərtəbə + b**

Hər feature-ın öz **w**-si var: "bu feature 1 vahid artsa, target nə qədər dəyişir?"

Kiçik nümunə: sahə **və** otaq sayı ilə qiymət.

````python
df_multi = pd.DataFrame({
    "sahe":  [35, 45, 50, 60, 65, 75, 85, 95, 110, 130],
    "otaq":  [1,  1,  2,  2,  2,  3,  3,  3,   4,   4],
    "qiymet":[75, 95, 100, 130, 125, 150, 175, 185, 215, 255]
})

X_multi = df_multi[["sahe", "otaq"]]
y_multi = df_multi["qiymet"]

model_multi = LinearRegression()
model_multi.fit(X_multi, y_multi)

print(f"w_sahe = {model_multi.coef_[0]:.2f}")
print(f"w_otaq = {model_multi.coef_[1]:.2f}")
print(f"b      = {model_multi.intercept_:.2f}")
print(f"\nFormula: qiymət = {model_multi.coef_[0]:.2f}×sahə + {model_multi.coef_[1]:.2f}×otaq + {model_multi.intercept_:.2f}")

# Yeni mənzil: 80 kv.m, 3 otaq
yeni_multi = pd.DataFrame({"sahe": [80], "otaq": [3]})
print("80 kv.m, 3 otaq ->", round(model_multi.predict(yeni_multi)[0]), "min AZN")
````

### 3️⃣ Polynomial Regression (Polinomial)

Bəzən mənzil qiyməti ilə sahə arasında əlaqə **düz xətt deyil** — əyri olur.

Məsələn: kiçik mənzillərdə sahə artdıqca qiymət mülayim artır, amma **böyük / premium** mənzillərdə (100+ kv.m) qiymət **daha sürətlə** yüksəlir.

Sadə fikir: sahənin özünü və **kvadratını** ($sahe^2$) modele veririk:

> **qiymət = w₁ × sahə + w₂ × sahə² + b**

sklearn-də bunu `PolynomialFeatures` ilə edirik — sonra yenə eyni `LinearRegression`.

````python
from sklearn.preprocessing import PolynomialFeatures

# Mənzillər: sahə böyüdükcə qiymət əyri (premium effekt) artır
df_poly = pd.DataFrame({
    "sahe":   [30, 40, 50, 60, 70, 85, 100, 120, 140],
    "qiymet": [55, 70, 90, 115, 145, 190, 250, 340, 460]  # min AZN
})

#region Polynominal learning
# dərəcə = 2 → sahə və sahə² sütunları yaradır
poly = PolynomialFeatures(degree=2, include_bias=False) # Qeyd edirəm ki, mənim 2 dərəcəli featurelarım var
X_poly = poly.fit_transform(df_poly[["sahe"]]) # sahəyə baxıb model öyrənir

model_poly = LinearRegression() # linear regression yaradıram
model_poly.fit(X_poly, df_poly["qiymet"]) # X_poly(artıq mənim sahələrimə uyğun hesablanmış polynominal xüsusiyyətklərdir) son model öyrənir

#endregion
print("Feature adları:", poly.get_feature_names_out(["sahe"]))
print(f"w_sahe  = {model_poly.coef_[0]:.4f}")
print(f"w_sahe2 = {model_poly.coef_[1]:.4f}")
print(f"b       = {model_poly.intercept_:.2f}")
print(f"\nFormula: qiymət = {model_poly.coef_[0]:.4f}×sahə + {model_poly.coef_[1]:.4f}×sahə² + {model_poly.intercept_:.2f}")

#region base Linear learning 
# Qrafik: düz xətt vs əyri
model_duz = LinearRegression().fit(df_poly[["sahe"]], df_poly["qiymet"])
sahe_grid = pd.DataFrame({"sahe": np.linspace(30, 140, 100)})

#endregion
plt.scatter(df_poly["sahe"], df_poly["qiymet"], s=70, color="black", label="Real mənzillər", zorder=3)
plt.plot(sahe_grid["sahe"], model_duz.predict(sahe_grid), "--", color="gray", label="Simple Linear")
plt.plot(sahe_grid["sahe"], model_poly.predict(poly.transform(sahe_grid)), color="blue", linewidth=2.5, label="Polynomial (dərəcə 2)")
plt.xlabel("Sahə (kv.m)")
plt.ylabel("Qiymət (min AZN)")
plt.title("Mənzil: düz xətt vs Polynomial əyri")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

#### Eyni ideya, amma 2 feature: sahə + otaq

Real həyatda yalnız sahə kifayət etmir. **Otaq sayı** da qiymətə təsir edir — və əlaqə yenə əyri ola bilər.

`PolynomialFeatures(degree=2)` iki feature üçün belə sütunlar yaradır:

| Feature | Mənası |
|---|---|
| `sahe` | sahə |
| `otaq` | otaq sayı |
| `sahe^2` | sahənin kvadratı |
| `sahe otaq` | sahə × otaq (birlikdə təsir) |
| `otaq^2` | otaq sayının kvadratı |

> **qiymət = w₁·sahə + w₂·otaq + w₃·sahə² + w₄·(sahə·otaq) + w₅·otaq² + b**

````python
# Mənzillər: sahə + otaq → qiymət (əyri əlaqə)
df_poly2 = pd.DataFrame({
    "sahe":   [30, 40, 50, 55, 65, 75, 90, 110, 130],
    "otaq":   [1,  1,  2,  2,  2,  3,  3,   4,   4],
    "qiymet": [55, 68, 95, 110, 135, 175, 230, 320, 430]  # min AZN
})

poly2 = PolynomialFeatures(degree=2, include_bias=False)
X_poly2 = poly2.fit_transform(df_poly2[["sahe", "otaq"]])

model_poly2 = LinearRegression()
model_poly2.fit(X_poly2, df_poly2["qiymet"])

print("Feature adları:", poly2.get_feature_names_out(["sahe", "otaq"]))
for ad, w in zip(poly2.get_feature_names_out(["sahe", "otaq"]), model_poly2.coef_):
    print(f"  {ad:12s} → w = {w:.4f}")
print(f"  {'b':12s} → {model_poly2.intercept_:.2f}")

# Yeni mənzil: 80 kv.m, 3 otaq
yeni = pd.DataFrame({"sahe": [80], "otaq": [3]})
proqnoz = model_poly2.predict(poly2.transform(yeni))[0]
print(f"\n80 kv.m, 3 otaq → prediction: {proqnoz:.0f} min AZN")

# Müqayisə: Multiple Linear (düz) vs Polynomial
model_multi2 = LinearRegression().fit(df_poly2[["sahe", "otaq"]], df_poly2["qiymet"])
p_duz = model_multi2.predict(df_poly2[["sahe", "otaq"]])
p_poly = model_poly2.predict(X_poly2)

from sklearn.metrics import r2_score
print(f"R² Multiple Linear : {r2_score(df_poly2['qiymet'], p_duz):.3f}")
print(f"R² Polynomial      : {r2_score(df_poly2['qiymet'], p_poly):.3f}")

# Qrafik: hər otaq sayı üçün hamar əyri (birinci nümunə kimi)
sahe_grid = np.linspace(30, 130, 100)
otaq_reng = {1: "#4C78A8", 2: "#F58518", 3: "#54A24B", 4: "#E45756"}

plt.figure(figsize=(8, 5))
plt.scatter(df_poly2["sahe"], df_poly2["qiymet"], s=90, color="black",
            label="Real mənzillər", zorder=4, edgecolors="white", linewidths=0.8)

for otaq_n, reng in otaq_reng.items():
    grid = pd.DataFrame({"sahe": sahe_grid, "otaq": otaq_n})
    # Multiple Linear — boz kəsik xətt
    plt.plot(sahe_grid, model_multi2.predict(grid), "--", color=reng, alpha=0.35, linewidth=1.5)
    # Polynomial — dolu əyri
    plt.plot(sahe_grid, model_poly2.predict(poly2.transform(grid)),
             color=reng, linewidth=2.5, label=f"Polynomial · {otaq_n} otaq")

# Real nöqtələrin yanında otaq sayını yazırıq
for _, row in df_poly2.iterrows():
    plt.annotate(f'{int(row["otaq"])}o', (row["sahe"], row["qiymet"]),
                 textcoords="offset points", xytext=(6, 4), fontsize=9, color="#333")

plt.plot([], [], "--", color="gray", label="Multiple Linear (kəsik)")
plt.xlabel("Sahə (kv.m)")
plt.ylabel("Qiymət (min AZN)")
plt.title("Mənzil: Multiple Linear vs Polynomial (sahə + otaq)")
plt.legend(loc="upper left", framealpha=0.9)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
````

| Bir neçə feature, hər biri təsir edir | **Multiple Linear** |
| Əlaqə əyri görünür (parabola və s.) | **Polynomial** |

> ⚠️ Polynomial-da dərəcəni çox artırmaq **overfitting** riski yaradır: model verilənləri "əzbərləyir", yeni nümunələrdə zəif işləyir.

# 🅱️ Hissə 2. Metrics

## 6️⃣ Metric nədir? (Concept of Metric)

**Metric = modelin keyfiyyətini ölçən ədəddir.**

Analogiya:
- Tələbənin biliyini **imtahan balı** ilə ölçürük
- Maşının sürətini **spidometr** ilə ölçürük
- Modelin keyfiyyətini **metric** ilə ölçürük

Metric bizə imkan verir:
- ✅ Modelin yaxşı olub-olmadığını bilmək
- ✅ İki modeli bir-biri ilə **müqayisə etmək**
- ✅ Modeli yaxşılaşdırdıqda irəliləyiş olub-olmadığını görmək

Regression üçün 3 məşhur metric öyrənəcəyik: **MAE**, **MSE**, **R²**.

````python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Əvvəlki modelin prediction-ları
p = model.predict(X)

# Hər mənzil üçün real, prediction və error
cedvel = pd.DataFrame({
    "real": y,
    "prediction": p.round(0),
    "error (real - prediction)": (y - p).round(0)
})
cedvel
````

### 🤔 Niyə error-ları sadəcə toplamaq olmaz?

Bəzi error-lar **müsbət** (model az dedi), bəzilər **mənfidir** (model çox dedi). Toplasaq bir-birini **ləğv edə** bilər.

````python
error = y - p
print("Error-ların cəmi:", round(error.sum(), 2))
print("Amma model mükəmməl deyil! Ən böyük error:", round(abs(error).max(), 1), "min AZN")
````

Cəm təxminən **0** çıxdı, amma model səhv edir. Deməli, error-ları toplamazdan əvvəl **müsbət etməliyik**. Bunun iki yolu var və bizə iki metric verir:

| Yol | Metric |
|---|---|
| Modulu götürmək (işarəni atmaq) | **MAE** |
| Kvadrata yüksəltmək | **MSE** |

## 7️⃣ MAE: Mean Absolute Error

**Mean Absolute Error = error-ların modulunun orta qiyməti.**

3 addım:
1. Hər mənzil üçün error tap
2. İşarəni at (absolute)
3. Orta qiyməti hesabla

````python
# Əl ilə
error = y - p              # 1. error
mutleq = abs(error)        # 2. işarəni atırıq
mae = mutleq.mean()        # 3. orta
print(f"MAE (əl ilə)  = {mae:.2f}")

# sklearn ilə (eyni nəticə)
print(f"MAE (sklearn) = {mean_absolute_error(y, p):.2f}")
````

### 💬 MAE-ni necə oxuyuruq?

> Modelimiz **orta hesabla MAE min AZN** səhv edir.

- ✅ Vahidi targetlə eynidir (min AZN), başa düşmək asandır
- ✅ Hər error-a bərabər əhəmiyyət verir
- 🎯 Nə qədər **kiçik**, o qədər yaxşı (ideal = 0)

## 8️⃣ MSE: Mean Squared Error

**Mean Squared Error = error-ların kvadratının orta qiyməti.**

3 addım:
1. Hər mənzil üçün error tap
2. Kvadrata yüksəlt
3. Orta qiyməti hesabla

**Niyə kvadrat?** Böyük səhvləri daha ağır cəzalandırır:

| Error | Kvadratı |
|---|---|
| 2 | 4 |
| 10 | 100 |

````python
# Əl ilə
kvadrat = error ** 2       # 2. kvadrata yüksəldirik
mse = kvadrat.mean()       # 3. orta
print(f"MSE (əl ilə)  = {mse:.2f}")

# sklearn ilə
print(f"MSE (sklearn) = {mean_squared_error(y, p):.2f}")

# Bonus: RMSE (kökü) vahidi yenə min AZN edir
print(f"RMSE = {mse ** 0.5:.2f}  (MSE-nin kvadrat kökü)")
````

### 💬 MSE-ni necə oxuyuruq?

- Vahidi **kvadratdır** (min AZN²), ona görə birbaşa şərh etmək çətindir. Bunun üçün **RMSE** (kvadrat kök) istifadə olunur.
- 🎯 Nə qədər **kiçik**, o qədər yaxşı (ideal = 0)
- Böyük error-lara çox həssasdır

### 🧪 Eksperiment: bir "səhv" qiymət (Outlier) nə edir?

Təsəvvür edin ki, bir mənzilin qiyməti **səhvən** 100 min artıq yazılıb. Bu cür qeyri-adi dəyərə **outlier** deyirik.
MAE və MSE necə reaksiya verəcək?

````python
y_outlier = y.copy()
y_outlier.iloc[3] += 100        # 4-cü mənzilin qiymətini səhvən 100 artırdıq

print("                  MAE      MSE")
print(f"Normal verilən: {mean_absolute_error(y, p):6.1f}  {mean_squared_error(y, p):7.1f}")
print(f"Outlier ilə:    {mean_absolute_error(y_outlier, p):6.1f}  {mean_squared_error(y_outlier, p):7.1f}")
````

📌 **Nəticə:**
- **MAE** təxminən 4 dəfə artdı (3.1 → 13.1)
- **MSE** təxminən 70 dəfə artdı (16 → 1171)

| | MAE | MSE |
|---|---|---|
| Outlier-ə həssaslıq | Az | Çox |
| Böyük səhvləri cəzalandırma | Bərabər | Güclü |
| Vahid | Targetlə eyni | Kvadrat |

Əgər verilənlərdə outlier çoxdursa, **MAE** daha etibarlıdır. Əgər böyük səhvlər xüsusilə təhlükəlidirsə, **MSE** seçilir.

## 9️⃣ R² (R squared) metric

MAE və MSE bizə səhvin **böyüklüyünü** deyir. Amma "50 min AZN səhv" yaxşıdır, yoxsa pis? Bu, verilənlərdən asılıdır.

**R² sual verir:** *Bizim model, həmişə **orta qiyməti** deyən "ağılsız" modeldən nə qədər yaxşıdır?*

### Ağılsız model
Hər mənzil üçün eyni cavab: **orta qiymət**. Qrafikdə bu, üfüqi xəttdir.

````python
orta = y.mean()

plt.scatter(df["sahe"], y, s=70, color="black", label="Real mənzillər")
plt.axhline(orta, color="gray", linestyle="--", label=f"Ağılsız model (hamısına {orta:.0f})")
plt.plot(x_xett, [w * x + b for x in x_xett], color="blue", linewidth=3, label="Bizim model")
plt.xlabel("Sahə (kv.m)")
plt.ylabel("Qiymət (min AZN)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

Bizim modelin xətti nöqtələrə **daha yaxındır**. R² bunu bir ədədlə ifadə edir:

> **R² = 1 − (bizim modelin error-u / ağılsız modelin error-u)**

````python
# Əl ilə
bizim_error = ((y - p) ** 2).sum()          # bizim modelin kvadrat error-ları
agilsiz_error = ((y - orta) ** 2).sum()     # "həmişə orta" modelinin error-ları
r2 = 1 - bizim_error / agilsiz_error
print(f"R² (əl ilə)  = {r2:.3f}")

# sklearn ilə
print(f"R² (sklearn) = {r2_score(y, p):.3f}")
````

### 💬 R²-ni necə oxuyuruq?

| R² | Mənası |
|---|---|
| **1.0** | Mükəmməl model |
| **0.9** | Model dəyişkənliyin təxminən 90%-ni izah edir, çox yaxşı |
| **0.0** | Model "həmişə orta" deyən modeldən yaxşı deyil |
| **< 0** | Model orta qiymətdən də pisdir 😬 |

- 🎯 Nə qədər **1-ə yaxın**, o qədər yaxşı (MAE və MSE-nin əksinə!)
- Vahidi yoxdur, müxtəlif tapşırıqlarda müqayisə etmək asandır

### 📋 Üç metric-in xülasəsi

| Metric | Nə edir? | Yaxşı nəticə | Vahid |
|---|---|---|---|
| **MAE** | Error-ların modulunun ortası | Kiçik | Target ilə eyni |
| **MSE** | Error-ların kvadratının ortası | Kiçik | Kvadrat |
| **R²** | Ağılsız modellə müqayisə | 1-ə yaxın | Vahidsiz |

## 🔟 Loss function və Cost function

Metrics bizə modelin keyfiyyətini **göstərir**. Amma kompüter w və b-ni tapanda onu **nə idarə edir?**

Cavab: **Loss function** və **Cost function**.

| Anlayış | Nəyi ölçür? | Nümunə |
|---|---|---|
| **Loss function** | **Bir** nümunə üzrə error | Bir mənzil üçün (real − prediction)² |
| **Cost function** (Error function) | **Bütün** nümunələr üzrə orta loss | Bütün mənzillərin loss-larının ortası |

> 💡 Bəzi mənbələrdə Cost function-a "Error function" da deyilir. Termin fərqli ola bilər, amma ideya eynidir.

````python
# Loss: yalnız bir mənzil üçün
i = 3
loss_bir = (y.iloc[i] - p[i]) ** 2
print(f"Mənzil #{i}: real={y.iloc[i]}, prediction={p[i]:.0f}, loss={loss_bir:.1f}")

# Cost: bütün mənzillər üçün loss-ların ortası
butun_loss = (y - p) ** 2
cost = butun_loss.mean()
print(f"Cost (bütün loss-ların ortası) = {cost:.2f}")
print(f"Müqayisə üçün MSE              = {mean_squared_error(y, p):.2f}")
````

👀 **Cost = MSE!** Yəni MSE həm metric, həm də Cost function kimi işləyə bilər.

### Metric vs Loss/Cost: fərq nədir?

- **Loss/Cost function** → **kompüter** üçündür: modeli öyrədərkən minimuma çatdırmağa çalışır
- **Metric** → **insan** üçündür: nəticəni başa düşmək və müqayisə etmək üçün

Bəzən ikisi eynidir (MSE), bəzən fərqlənir.

### 🥣 Cost function-un forması

Gəlin w-nin müxtəlif qiymətləri üçün **Cost**-a baxaq (b-ni sabit saxlayırıq).

````python
import numpy as np

w_deyerleri = np.linspace(0, 4, 100)       # w üçün müxtəlif cəhdlər
cost_deyerleri = []
for w_cehd in w_deyerleri:
    texmin = w_cehd * df["sahe"] + b       # b sabit
    cost_deyerleri.append(((y - texmin) ** 2).mean())

en_yaxsi = w_deyerleri[np.argmin(cost_deyerleri)]

plt.plot(w_deyerleri, cost_deyerleri)
plt.scatter([en_yaxsi], [min(cost_deyerleri)], color="red", s=100, zorder=3, label=f"Minimum: w ≈ {en_yaxsi:.2f}")
plt.xlabel("w (weight)")
plt.ylabel("Cost")
plt.title("Cost function: w dəyişdikcə error necə dəyişir?")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

### 🎯 Əsas fikir

- Cost function **qab (parabola)** formasındadır
- Ən aşağı nöqtə = **ən kiçik error** = **ən yaxşı w**
- Modeli öyrətmək = bu qabın **dibini tapmaq**

Modelin tapdığı w ≈ **1.88** idi. Qrafikdəki minimum da ona uyğundur ✅

**Növbəti dərsdə** kompüterin bu qabın dibinə necə düşdüyünü öyrənəcəyik: **Gradient Descent** 🏔️

## 🧠 Dərsin yekunu

**ML-in 3 növü**
- **Supervised**: Features + Target (bu dərs — Regression)
- **Unsupervised**: yalnız Features, qrup/struktur tapır
- **Reinforcement**: mükafat/cəza ilə sına-yanıl öyrənir

**Regression**
- **Regression** Target-in *ədəd* olduğu tapşırıqdır (qiymət, temperatur, bal)
- Simple Linear Regression = verilənlərə **düz xətt** uyğunlaşdırmaq: `y = w × x + b`
- w (weight) və b (bias) tapmaq = modelin **öyrənməsi** (training)
- sklearn-də 3 addım: `LinearRegression()` → `fit` → `predict`
- Digər modellər: **Multiple Linear**, **Polynomial**

**Metrics**
- **Metric** modelin keyfiyyətini ölçən ədəddir
- **MAE**: orta |error|, outlier-ə az həssas, vahidi target ilə eynidir
- **MSE**: orta error², böyük səhvləri güclü cəzalandırır
- **R²**: ağılsız modellə müqayisə, 1-ə yaxın yaxşıdır
- **Loss function**: bir nümunənin error-u. **Cost function**: bütün loss-ların ortası
- Modeli öyrətmək = Cost function-un minimumunu tapmaq

def my_mae(y_real, y_pred):
    return np.mean(np.abs(y_real - y_pred))

def my_mse(y_real, y_pred):
    return np.mean((y_real - y_pred) ** 2)

def my_r2(y_real, y_pred):
    bizim_error = np.sum((y_real - y_pred) ** 2)
    agilsiz_error = np.sum((y_real - np.mean(y_real)) ** 2)
    return 1 - bizim_error / agilsiz_error
